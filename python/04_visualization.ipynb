```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
root = Path.cwd()

data_path = root.parent / "data" / "cleaned" / "Superstore_Cleaned.csv"

df = pd.read_csv(data_path)

df.head()
```

```python
sales_category = df.groupby("Category")["Sales"].sum().sort_values()

plt.figure(figsize=(8,5))
sales_category.plot(kind="bar")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

```python
profit_category = df.groupby("Category")["Profit"].sum().sort_values()

plt.figure(figsize=(8,5))
profit_category.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum()

plt.figure(figsize=(8,5))
region_profit.plot(kind="bar")

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()
```

```python
sales_category = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(7,7))

sales_category.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Sales Distribution by Category")
plt.ylabel("")

plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.index = monthly_sales.index.astype(str)
```

```python
plt.figure(figsize=(12,5))

monthly_sales.plot(kind="line")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
top_customer = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

top_customer.plot(kind="bar")

plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer")
plt.ylabel("Sales")

plt.tight_layout()

plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

top_products.plot(kind="bar")

plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Discount"], bins=20)

plt.title("Discount Distribution")
plt.xlabel("Discount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
sales_category = df.groupby("Category")["Sales"].sum().sort_values()

plt.figure(figsize=(8,5))
sales_category.plot(kind="bar")

plt.title("Sales by Category")

plt.tight_layout()

plt.savefig("../images/sales_by_category.png")

plt.show()
```
